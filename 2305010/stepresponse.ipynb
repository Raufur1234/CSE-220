```python
def step_response_from_running_sum(h):
    s = DiscreteSignal(h.start_time, h.end_time)
    running_total = 0.0
    for n in h.times():
        running_total += h.get_value_at_time(n)
        s.set_value_at_time(n, running_total)
    return s


h = make_signal(0, 2, [1.0, 0.5, 0.25])
s_direct = step_response_from_running_sum(h)
print(s_direct.values)   # -> [1.0, 1.5, 1.75]

# cross-check against the Q7 method
u = make_unit_step(-3, 8)
s_via_lti = LTISystem(h).output(u)
for n in range(0, 3):
    print(n, s_direct.get_value_at_time(n), s_via_lti.get_value_at_time(n))
# 0 1.0 1.0
# 1 1.5 1.5
# 2 1.75 1.75   -- exact match

# non-causal case
h2 = make_signal(-2, 1, [0.3, -0.2, 1.0, 0.5])
s2_direct = step_response_from_running_sum(h2)
u2 = make_unit_step(-6, 8)
s2_via_lti = LTISystem(h2).output(u2)
for n in range(-2, 2):
    print(n, s2_direct.get_value_at_time(n), s2_via_lti.get_value_at_time(n))
# -2 0.3  0.3
# -1 0.1  0.1
#  0 1.1  1.1
#  1 1.6  1.6   -- also exact match
```
